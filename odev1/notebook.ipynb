```python
import matplotlib.pyplot as plt 
```

```python
tablo=df.head()
```

```python
fig, ax=plt.subplots(figsize=(10,3))
ax.axis('tight')
ax.axis('off')

tabl= ax.table(
    cellText=tablo.values,
    colLabels=tablo.columns,
    cellLoc='center',
    loc='center'
)
tabl.auto_set_font_size(False)
tabl.set_fontsize(12)
tabl.scale(1,2)

plt.show()
```

```python

# Her sütun için sırayla grafik oluştur
for sutun in df.columns:
    plt.figure(figsize=(8, 4))
    
    # 1. Durum: Kategorik veya Metin tipi sütunlar
    if df[sutun].dtype == 'object' or df[sutun].nunique() < 15:
        # En çok tekrar eden ilk 10 değeri al
        deger_sayilari = df[sutun].value_counts().head(10)
        
        barlar = plt.bar(deger_sayilari.index.astype(str), deger_sayilari.values, color='#4C72B0', edgecolor='black')
        plt.title(f"Sütun Dağılımı: {sutun} (Adet Bazlı)", fontsize=12, fontweight='bold')
        plt.ylabel("Kişi / Adet Sayısı")
        
        # Çubukların üzerine net sayıları yaz
        plt.bar_label(barlar, padding=3)

    # 2. Durum: Sayısal (int/float) sütunlar (Aralık bazlı çubuk grafik)
    else:
        # Veriyi 10 eşit aralığa bölüp her aralıktaki sayıyı hesapla
        
        
        plt.title(f"Sütun Dağılımı: {sutun} (Aralık Bazlı Dağılım)", fontsize=12, fontweight='bold')
        plt.ylabel("Frekans (Gözlem Sayısı)")

    # Genel grafik düzenlemeleri
    plt.xlabel(sutun, fontweight='bold')
    plt.xticks(rotation=30, ha='right')
    plt.grid(axis='y', linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.show()
```
